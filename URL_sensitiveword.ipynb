```python
import pandas as pd
```

```python
df = pd.read_csv('data/raw/url_binary_dataset.csv')

df.head()
```

```python
malicious_urls = df.loc[df["status"] == "악성", "url"].astype(str)
```

```python
tokens = (
    malicious_urls
    .str.lower()
    .str.split(r"[.\-/]", regex=True)
    .explode()
    .str.strip()
)
```

```python
token = tokens[
    tokens.ne("")
    & ~tokens.isin(["http:", "https:", "www"])
]
```

```python
token_counts = tokens.value_counts()
print(token_counts.head(20))
```

```python
clean_urls = df.loc[df["status"] == "정상", "url"].astype(str)
clean_tokens = (
    clean_urls
    .str.lower()
    .str.split(r"[.\-/]", regex=True)
    .explode()
    .str.strip()
)
clean_token = clean_tokens[
    clean_tokens.ne("")
    & ~clean_tokens.isin(["http:", "https:", "www"])
]
clean_token_counts = clean_tokens.value_counts()
print(clean_token_counts.head(20))
```

```python
# 정상 토큰 이름 집합
clean_token_set = set(clean_tokens.dropna())

# 악성 토큰 이름 집합
malicious_token_set = set(token_counts.index)

# 정상과 악성에 모두 존재하는 토큰
common_tokens = clean_token_set & malicious_token_set

# token_counts의 인덱스를 기준으로 공통 토큰 제외
malicious_only_counts = token_counts[
    ~token_counts.index.isin(common_tokens)
]

print("\n악성 URL에만 존재하는 토큰:")
print(malicious_only_counts.head(20))
```

```python
print("\n악성 URL에만 존재하는 토큰:")
malicious_only_result = (
    malicious_only_counts
    .head(50)
    .rename_axis("token")
    .reset_index(name="count")
)

print(malicious_only_result)
```
